```
import pandas as pd

# Load Heart Disease Dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"

df = pd.read_csv(url, header=None)

print(df.head())

# Unique values in target column
print(set(df[13]))

# Convert target values (2,3,4) into 1
df[13].replace(to_replace=[2, 3, 4], value=1, inplace=True)

print(df.head())
print(set(df[13]))

# Binarizing MNIST Dataset
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', version=1, as_frame=False)

print(mnist.data.shape)
print(mnist.target.shape)


import matplotlib.pyplot as plt

# Display 8 images
for i, (img, y) in enumerate(zip(mnist.data[15:23], mnist.target[15:23])):
    plt.subplot(2, 4, i + 1)
    plt.imshow(img.reshape(28, 28), cmap='gray')
    plt.xticks([])
    plt.yticks([])
    plt.title(y)

plt.savefig('mnist_samples.png', bbox_inches='tight', dpi=350)
plt.show()


# Print first image
print(mnist.data[0].reshape(28, 28))

# Print small 8x8 portion
print(mnist.data[0].reshape(28, 28)[10:18, 10:18])


# Convert grayscale pixels into binary pixels
mnist.data[mnist.data < 128] = 0
mnist.data[mnist.data >= 128] = 1

print(mnist.data[0].reshape(28, 28)[10:18, 10:18])


# Binary Classification (Only digits 7 and 8)
X = mnist.data[(mnist.target == '7') | (mnist.target == '8')]
y = mnist.target[(mnist.target == '7') | (mnist.target == '8')]

print(X.shape)
print(y.shape)

print(y[:10])

# Convert labels
y = [0 if v == '7' else 1 for v in y]

print(y[:10])
```
