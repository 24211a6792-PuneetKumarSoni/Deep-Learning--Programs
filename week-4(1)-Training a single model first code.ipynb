```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_classes=2,
    n_informative=2,
    n_redundant=0,
    n_repeated=0,
    n_clusters_per_class=1,
    class_sep=1.5,
    random_state=5
)

y[y == 0] = -1

fig = plt.figure()

plt.plot(X[0,0], X[0,1], 'rx', label='$-1$ - Negative')
plt.plot(X[1,0], X[1,1], 'bd', label='$+1$ - Positive')

for i in range(len(y)):
    x1 = X[i,0]
    x2 = X[i,1]

    if (y[i] == -1):
        plt.plot(x1, x2, 'rx')
    else:
        plt.plot(x1, x2, 'bd')

plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.legend()
plt.title('Sample two-dimensional data')
plt.axis('tight')
plt.savefig('ch.5.two-class.data.png', dpi=350, bbox_inches='tight')
plt.savefig('ch.5.two-class.data.eps', dpi=350, bbox_inches='tight')
plt.show()
```
